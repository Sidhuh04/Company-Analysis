```python
bins = [0, 20, 30, 40, 50, 60]
labels = ['Below 20', '20-30', '30-40', '40-50', '50-60']

df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels)
```

```python
age_count = df['Age_Group'].value_counts().sort_index()
print(age_count)
```

```python
age_count = df['Age_Group'].value_counts().sort_index()
print(age_count)
```

```python
print("Predominant Age Group:", age_count.idxmax())
print("Number of Employees:", age_count.max())
```

```python
team_salary = df.groupby('Team')['Salary'].sum().sort_values(ascending=False)
position_salary = df.groupby('Position')['Salary'].sum().sort_values(ascending=False)
```

```python
print("Team with highest salary expenditure:")
print(team_salary.idxmax(), "is", team_salary.max())

print("Position with highest salary expenditure:")
print(position_salary.idxmax(), "is", position_salary.max())
```

```python
correlation = df['Age'].corr(df['Salary'])

print("Correlation between Age and Salary:", correlation)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.scatterplot(data=df, x='Age', y='Salary')

plt.title('Correlation between Age and Salary')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.show()
```
